# PrefixSum(Different Versions)

```python
def prefix_sum(arr):
    """Compute prefix sum array."""
    if not arr:
        return []

    prefix = [0] * len(arr)
    prefix[0] = arr[0]

    for i in range(1, len(arr)):
        prefix[i] = prefix[i-1] + arr[i]

    return prefix


# Example usage
if __name__ == "__main__":
    nums = [1, 2, 3, 4, 5]
    print("Original:", nums)
    print("Prefix Sum:", prefix_sum(nums))
    # Output: [1, 3, 6, 10, 15]
```

```python
def prefix_sum_inplace(arr):
    """Compute prefix sum in-place."""
    for i in range(1, len(arr)):
        arr[i] += arr[i-1]
    return arr


# Example
nums = [1, 2, 3, 4, 5]
print(prefix_sum_inplace(nums))  # [1, 3, 6, 10, 15]
```

```python
from itertools import accumulate

nums = [1, 2, 3, 4, 5]
prefix = list(accumulate(nums))
print(prefix)  # [1, 3, 6, 10, 15]
```

```python
class PrefixSum:
    def __init__(self, arr):
        self.prefix = [0] * (len(arr) + 1)
        for i in range(1, len(arr) + 1):
            self.prefix[i] = self.prefix[i-1] + arr[i-1]

    def query(self, left, right):
        """Get sum from index left to right (inclusive)"""
        return self.prefix[right+1] - self.prefix[left]


# Example
ps = PrefixSum([1, 2, 3, 4, 5])
print(ps.query(1, 3))  # Sum from index 1 to 3 → 9 (2+3+4)
```
